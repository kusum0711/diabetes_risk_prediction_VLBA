#### 1.5 Data Requirements – Needed Data & Why

### Data Requirements

To predict diabetes risk, the following types of data were required:

1. **Medical Indicators**
   - High blood pressure
   - High cholesterol
   - BMI
   - Stroke history
   - Heart disease history

2. **Lifestyle Factors**
   - Smoking
   - Alcohol consumption
   - Physical activity
   - Fruit and vegetable intake

3. **Healthcare Access**
   - Insurance coverage
   - Doctor visit affordability

4. **Demographic Factors**
   - Age
   - Gender
   - Education
   - Income

### Why Needed?

These variables are strongly associated with diabetes development and are useful predictors for classification models.

#### 3.2 BIVARIATE ANALYSIS

The purpose of bivariate analysis is to study relationships between predictor variables and the target variable (Diabetes_012).

This helps identify strong diabetes indicators and relevant features for model training.

##### Continuous Features

```python


fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes = axes.flatten()

CLASS_PALETTE = ['#378ADD', '#EF9F27', '#E24B4A']
CLASS_LABELS  = {0.0: 'No diabetes', 1.0: 'Prediabetes', 2.0: 'Diabetes'}

for i, col in enumerate(['BMI', 'MentHlth', 'PhysHlth']):
    for cls, color in zip([0.0, 1.0, 2.0], CLASS_PALETTE):
        subset = df[df['Diabetes_012'] == cls][col]
        sns.kdeplot(subset, ax=axes[i], color=color,
                    linewidth=2, fill=True, alpha=0.25,
                    label=CLASS_LABELS[cls])

    axes[i].set_xlim(df[col].min(), df[col].quantile(0.99))
    axes[i].set_title(f'{col} by class', fontweight='bold')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Density')
    axes[i].legend(fontsize=9)

plt.suptitle('Continuous Feature distribution by diabetes class',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

##### Ordinal Features

```python

fig, axes = plt.subplots(2, 2, figsize=(14, 8))
axes = axes.flatten()

CLASS_PALETTE = ['#378ADD', '#EF9F27', '#E24B4A']
CLASS_LABELS  = {0.0: 'No diabetes', 1.0: 'Prediabetes', 2.0: 'Diabetes'}

ordinal_cols = ['GenHlth', 'Age', 'Education', 'Income']

for i, col in enumerate(ordinal_cols):
    for cls, color in zip([0.0, 1.0, 2.0], CLASS_PALETTE):
        subset = df[df['Diabetes_012'] == cls][col]

        sns.kdeplot(
            subset,
            ax=axes[i],
            color=color,
            linewidth=2,
            fill=True,
            alpha=0.25,
            label=CLASS_LABELS[cls]
        )

    axes[i].set_title(f'{col} vs Diabetes', fontweight='bold')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Density')
    axes[i].legend(fontsize=9)

plt.suptitle('Ordinal features distribution by diabetes class',
             fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()
```

##### Binary Features

```python

# Simple comparison: Which group has more diabetes?

# Binary columns to compare
features = [
    'HighBP', 'HighChol', 'Smoker', 'Stroke',
    'HeartDiseaseorAttack', 'PhysActivity',
    'Fruits', 'Veggies', 'DiffWalk'
]

results = []

for col in features:
    # Diabetes rate among people with value = 1
    yes_rate = (df[df[col] == 1]['Diabetes_012'] == 2).mean() * 100
    
    # Diabetes rate among people with value = 0
    no_rate = (df[df[col] == 0]['Diabetes_012'] == 2).mean() * 100
    
    results.append([col, yes_rate, no_rate])

# Create dataframe
compare_df = pd.DataFrame(results, columns=['Feature', 'Yes(1)%', 'No(0)%'])
compare_df.set_index('Feature', inplace=True)

# Plot (same style)
compare_df.plot(kind='bar', figsize=(12,6), color=['red','gray'])

plt.title("Diabetes % in Yes vs No Groups")
plt.ylabel("Diabetes Percentage")
plt.xticks(rotation=45)
plt.legend(['Yes', 'No'])
plt.show()
```

# 4: Data Quality Report

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/diabetes.csv")

print("Shape:", df.shape)
df.head()
```

MISSING VALUES

```python
print("Missing Values:")
df.isnull().sum()
```

DUPLICATES

```python
dup_count = df.duplicated().sum()
dup_pct = (dup_count / len(df)) * 100

print("Total duplicates:", dup_count)
print("Percentage:", round(dup_pct, 2), "%")

df[df.duplicated()].head()
```

DATA CLEANING

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after cleaning:", df.shape)
```

FEATURE ENGINEERING (BMI)

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 0
    elif bmi < 25:
        return 1
    elif bmi < 30:
        return 2
    else:
        return 3

df['BMI_cat'] = df['BMI'].apply(bmi_category)

df['BMI_cat'].value_counts()
```

FEATURE ENGINEERING (AGE)

```python
def age_category(age):
    if age <= 4:
        return 0   # Young
    elif age <= 8:
        return 1   # Middle age
    else:
        return 2   # Older

df['Age_cat'] = df['Age'].apply(age_category)

df['Age_cat'].value_counts()
```

BMI Category vs Diabetes

```python
import seaborn as sns

plt.figure(figsize=(7,5))

sns.countplot(x='BMI_cat', hue='Diabetes_012', data=df)

plt.title("Diabetes Distribution Across BMI Categories")
plt.xlabel("BMI Category (0=Under, 1=Normal, 2=Over, 3=Obese)")
plt.ylabel("Count")

plt.show()
```

The relationship between BMI categories and diabetes status was analyzed. 
The results show that individuals in higher BMI categories (overweight and obese) 
have a significantly higher frequency of diabetes cases, indicating a strong 
association between obesity and diabetes risk.

OUTLIER DETECTION

```python
numerical_cols = ['BMI', 'Age', 'MentHlth', 'PhysHlth']

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col}")
    print("Outliers:", len(outliers))
    print("Lower:", round(lower,2), "Upper:", round(upper,2))
```
